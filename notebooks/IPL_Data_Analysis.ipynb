# IPL Match & Player Performance Analysis

Place `matches.csv` and `deliveries.csv` inside `../dataset/` before running.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from src.data_cleaning import load_datasets, clean_datasets, save_cleaned
from src.exploratory_data_analysis import run_eda
```

## 1. Load and inspect data

```python
matches, deliveries = load_datasets()
print("Matches:", matches.shape, "Deliveries:", deliveries.shape)
display(matches.head())
display(deliveries.head())
```

## 2. Missing values and cleaning

```python
display(matches.isna().sum().sort_values(ascending=False).head(15))
display(deliveries.isna().sum().sort_values(ascending=False).head(15))
matches_clean, deliveries_clean = clean_datasets(matches, deliveries)
save_cleaned(matches_clean, deliveries_clean)
```

## 3. Exploratory analysis and charts

```python
run_eda(matches_clean, deliveries_clean)
```

## 4. Top run scorers

```python
batter = next((c for c in ["batter", "batsman"] if c in deliveries_clean), None)
runs = next((c for c in ["batsman_runs", "batter_runs"] if c in deliveries_clean), None)
if batter and runs:
    display(deliveries_clean.groupby(batter)[runs].sum().sort_values(ascending=False).head(10).to_frame("runs"))
else:
    print("Batter/run columns not found. Check your dataset schema.")
```
